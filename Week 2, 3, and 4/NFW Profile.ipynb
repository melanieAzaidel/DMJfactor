In this file, you'll be creating a plot of the NFW profile that we discussed. As a reminder, here's the function you're plotting.

$$
\rho(r) = \frac{\rho_0}{\frac{r}{R_s}\left(1 + \frac{r}{R_s}\right)^2}
$$

where
* $\rho(r)$ is the radial profile of dark matter in units of $\text{ GeV cm}^{-3}$,
* $r$ is the radial component within the Milky Way galaxy,
* $\rho_0 = \rho_\mathrm{sc} = 0.3\text{ GeV cm}^{-3}$ is the local energy density of dark matter at the solar circle, and
* $R_s = R_\mathrm{mw} = 20 \text{ kpc}$ is the radius of the Milky Way galaxy.

Your first task is to define a function that takes inputs and then produces an output. I will get you started with the skeleton of the function. 

One challenge you may come across is dealing with division by zero (try setting $r = 0$ in the above function and see what happens). You'll need to come up with an idea about how to alleviate this.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def nfw_density(r_kpc, rs_kpc=20, rho_0=0.3):
    """
    NFW dark matter density profile

    Parameters
    ----------
    r_kpc : float or array-like
        Radius in kpc (Galactocentric)
    rs_kpc : float
        Default scale radius in kpc
    rho_0 : float
        Default desired local density in GeV/cm^3

    Returns
    -------
    rho : float or ndarray
        Density in GeV/cm^3
    """
    
    rho = rho_0 * (rs_kpc / r_kpc) * (1 + r_kpc / rs_kpc)**(-2)

    return rho
```

Test your function below. The Sun is located 8.5 kpc away from the Galactic Center, so the following line of code should return the local dark matter density.

```python
local_density = nfw_density(8.5)
print(f"Local dark matter density at 8.5 kpc: {local_density:.3f} GeV/cm^3")
```
